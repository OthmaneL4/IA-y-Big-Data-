# 2. **Pyplot**
Con el módulo de pyplot podemos crear una figura, un área de trazado en una figura, trazar líneas en un área de trazado, decorar la trama con etiquetas, etc.

```python
import matplotlib.pyplot as plt

# inicializamos los datos
x = [10, 20, 30, 40]
y = [20, 25, 35, 55]

# plotting the data
plt.plot(x, y)

plt.show()
```

```python
# Añadimos título al gráfico
x = [10, 20, 30, 40]
y = [20, 25, 35, 55]
plt.plot(x, y)
plt.title("Gráfico lineal", fontsize=25, color="green")

plt.show()
```

```python
# Añadir etiquetas en los ejes x e y
x = [10, 20, 30, 40]
y = [20, 25, 35, 55]

plt.plot(x, y)

plt.title("Gráfico lineal", fontsize=25, color="green")

# Etiqueta en el eje y
plt.ylabel('Eje Y')

# Etiqueta en el eje x
plt.xlabel('Eje X')

plt.show()
```

```python
# Podemos añadir límites y etiquetas de graduación
x = [10, 20, 30, 40]
y = [20, 25, 35, 55]

plt.plot(x, y)

plt.title("Gráfico lineal", fontsize=25, color="green")

# Etiqueta en el eje y
plt.ylabel('Eje Y')

# Etiqueta en el eje x
plt.xlabel('Eje X')

# Límite en el eje Y
plt.ylim(0, 80)

# Etiquetas en el eje X
plt.xticks(x, labels=["uno", "dos", "tres", "cuatro"])

plt.show()
```

```python
# Añadir leyenda
x = [10, 20, 30, 40]
y = [20, 25, 35, 55]

plt.plot(x, y)

plt.title("Gráfico lineal", fontsize=25, color="green")

# Etiqueta en el eje y
plt.ylabel('Eje Y')

# Etiqueta en el eje x
plt.xlabel('Eje X')

# Límite en el eje Y
plt.ylim(0, 80)

# Etiquetas en el eje X
plt.xticks(x, labels=["uno", "dos", "tres", "cuatro"])

plt.legend(["GFG"])

plt.show()
```

```python
from matplotlib.figure import Figure
# La clase figura es la ventana donde se dibuja el gráfico.
x = [10, 20, 30, 40]
y = [20, 25, 35, 55]

fig = plt.figure(figsize =(7, 5), facecolor='g',
                 edgecolor='r', linewidth=7)

# Creamos una ventana de color verde con el borde rojo
ax = fig.add_axes([1, 1, 1, 1])
ax.plot(x, y)
plt.title("Gráfico lineal", fontsize=25, color="yellow")
# Etiqueta en el eje y
plt.ylabel('Eje Y')
# Etiqueta en el eje x
plt.xlabel('Eje X')
# Límite en el eje Y
plt.ylim(0, 80)
# Etiquetas en el eje X
plt.xticks(x, labels=["uno", "dos", "tres", "cuatro"])
plt.legend(["GFG"])

plt.show()
```

```python
# Usando el método subplot podemos agregar otro gráfico en la posición de cuadrícula especificada en la figura actual.
# subplot(numrows, numcols, fignum)
import matplotlib.pyplot as plt

x = [10, 20, 30, 40]
y = [20, 25, 35, 55]


plt.figure()

# Primer subgráfico
plt.subplot(121)
plt.plot(x, y)

# Segundo subgráfico
plt.subplot(122)
plt.plot(y, x)
```

```python
x = [0,1,2,3,4]
y = [0,2,4,6,8]

# Redimensión del gráfico (dpi especifica pixels en el momento de guardar la figura)
plt.figure(figsize=(8,5), dpi=100)


# Cambiamos argumentos para cambiar la visualización
plt.plot(x,y, label='2x', color='red', linewidth=2, marker='.', linestyle='--', markersize=10, markeredgecolor='blue');
```

```python
# Notación corta
#fmt es el marcador a usar y el estilo de línea mediante los parámetros color, marca y estilo de línea
#fmt = '[color][marker][line]'
plt.plot(x,y, 'b^--', label='2x')
plt.show()
```

# 2. **Clase de ejes**
La clase AXES es la unidad más básica y flexible para crear subparcelas.

Igual que la clase pyplot, la clase axes también proporciona métodos para agregar títulos, leyendas, límites, etiquetas, etc. Veamos algunos de ellos:

- Agregar título - ax.set_title()
- Agregar etiqueta X y etiqueta Y - ax.set_xlabel() , ax.set_ylabel()
- Establecer límites: ax.set_xlim() , ax.set_ylim()
- Etiquetas de marca: ax.set_xticklabels() , ax.set_yticklabels()
- Adición de leyendas: ax.legend()




```python
import matplotlib.pyplot as plt
from matplotlib.figure import Figure


x = [10, 20, 30, 40]
y = [20, 25, 35, 55]

fig = plt.figure(figsize = (5, 4))

# Añadimos los ejes a la figura
ax = fig.add_axes([1, 1, 1, 1])

# Dibujamos el primer conjunto
ax1 = ax.plot(x, y)

# Dibujamos el segundo conjunto
ax2 = ax.plot(y, x)

ax.set_title("Gráfico")


ax.set_xlabel("X")
ax.set_ylabel("Y")

# Adding Legend
ax.legend(labels = ('linea 1', 'linea 2'))

plt.show()
```

```python
# Utilizando patrones
labels = ['A', 'B', 'C']
values = [1,4,2]

plt.figure(figsize=(5,3), dpi=100)

bars = plt.bar(labels, values)

patterns = ['X', 'O', '*']
for bar in bars:
    bar.set_hatch(patterns.pop(0))

plt.show()
```

## 3. Histograma

Es un tipo de gráfico de barras en el que el eje X representa los rangos de intervalo, mientras que el eje Y proporciona información sobre la frecuencia.
Personalización que está disponible para el histograma:

- bins: número de bins de igual ancho
- color: para cambiar el color de la cara
- edgecolor: Color de los bordes
- estilo de línea: Para los edgelines
- alfa: valor de fusión, entre 0 (transparente) y 1 (opaco)

```python
x = np.random.normal(170, 10, 250)

plt.hist(x)
plt.show()
```

```python

# inicializamos los datos
x = data['petal_length']

# Dibujar el gráfico
plt.hist(x, bins=5, color='green', edgecolor='blue',
         linestyle='--', alpha=0.5)

# Añadir el título
plt.title("Iis")

# Etiqueta eje Y
plt.ylabel('Frecuencia')

# Etiqueta eje X
plt.xlabel('longitud pétalo')

plt.show()
```

## 4. Gráfico de dispersión

se utilizan para observar relaciones entre variables. El método scatter() en la biblioteca matplotlib se usa para dibujar un diagrama de dispersión.
Las personalizaciones que están disponibles para el diagrama de dispersión son:

- s: tamaño del marcador (puede ser escalar o una matriz de tamaño igual al tamaño de xoy)
- c: color de secuencia de colores para marcadores
- marcador: estilo de marcador
- anchos de línea: ancho del borde del marcador
- edgecolor: color del borde del marcador
- alfa: valor de fusión, entre 0 (transparente) y 1 (opaco)

```python
import numpy as np
x = np.array([5,7,8,7,2,17,2,9,4,11,12,9,6])
y = np.array([99,86,87,88,111,86,103,87,94,78,77,85,86])

plt.scatter(x, y)
plt.show()
```

```python
# Vamos a ver que relación existe entre dos variables, la velocidad (Y) y la edad (X).

#primera muestra, edad y velocidad de 13 coches:
x = np.array([5,7,8,7,2,17,2,9,4,11,12,9,6])
y = np.array([99,86,87,88,111,86,103,87,94,78,77,85,86])
plt.scatter(x, y)

#segunda muestra, edad y velocidad de 15 coches:
x = np.array([2,2,8,1,15,8,12,9,7,3,11,4,7,14,12])
y = np.array([100,105,84,105,90,99,90,95,94,100,79,112,91,80,85])
plt.scatter(x, y)

# Añadir el título
plt.title("coches")

# Etiqueta eje Y
plt.ylabel('Velocidad')

# Etiqueta eje X
plt.xlabel('Edad')
plt.show()
```

**Conclusión:** cuanto más nuevo es el coche, más velocidad puede alcanzar.

## 5. Gráfico circular

Se utiliza para mostrar solo una serie de datos.

Las personalizaciones que están disponibles para el gráfico circular son:

- explotar: moviendo las cuñas de la trama
- autopct: etiqueta la cuña con su valor numérico.
- color: el atributo se utiliza para dar color a las cuñas.
- sombra: se utiliza para crear la sombra de la cuña.

```python
cars = ['AUDI', 'BMW', 'FORD',
        'TESLA', 'JAGUAR',]
data = [23, 10, 35, 15, 12]

# Dibuja gráfico
plt.pie(data, labels=cars)

# Título del gráfico
plt.title("Car data")

plt.show()
```

```python
cars = ['AUDI', 'BMW', 'FORD',
        'TESLA', 'JAGUAR',]
data = [23, 13, 35, 15, 12]

explode = [0.1, 0.5, 0, 0, 0]

colors = ( "orange", "cyan", "yellow",
          "grey", "green",)

# Dibujo del gráfico
plt.pie(data, labels=cars, explode=explode, autopct='%1.2f%%',
        colors=colors, shadow=True)

plt.show()
```

## 7. Violín

```python
import matplotlib.pyplot as plt
import numpy as np


# make data:
np.random.seed(10)
D = np.random.normal((3, 5, 4), (0.75, 1.00, 0.75), (200, 3))

# plot:
fig, ax = plt.subplots()

vp = ax.violinplot(D, [2, 4, 6], widths=2,
                   showmeans=False, showmedians=False, showextrema=False)
# styling:
for body in vp['bodies']:
    body.set_alpha(0.9)
ax.set(xlim=(0, 8), xticks=np.arange(1, 8),
       ylim=(0, 8), yticks=np.arange(1, 8))

plt.show()
```
