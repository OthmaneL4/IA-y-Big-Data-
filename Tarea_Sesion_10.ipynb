
##  2. Fuente 1: JSON desde URL pública  
Carga el dataset de vehículos.


```python

url_json = "https://raw.githubusercontent.com/vega/vega-datasets/master/data/flights-2k.json"

df_vehicles = pd.read_json(url_json)

display(df_vehicles.head())
```


##  3. Fuente 2: API pública  
Usa una API relacionada con transporte, movilidad o tráfico.


```python

import requests
import pandas as pd

url_api = "https://api.citybik.es/v2/networks"

respuesta = requests.get(url_api, timeout=20)
respuesta.raise_for_status()

datos = respuesta.json()

df_bicicletas = pd.json_normalize(datos["networks"])

display(df_bicicletas.head())
```

```python
path = "/content/auto-mpg.csv"
df_auto = pd.read_csv(path)
display(df_auto.head())
```


##  6. Fuente 5: Web Scraping  
Obtén datos desde una página web que los publique en forma de tabla.




```python
import pandas as pd
import requests
from bs4 import BeautifulSoup
from io import StringIO

url = "https://en.wikipedia.org/wiki/List_of_countries_by_motor_vehicle_production"

headers = {
    "User-Agent": "Mozilla/5.0"
}

respuesta = requests.get(url, headers=headers, timeout=20)
respuesta.raise_for_status()

soup = BeautifulSoup(respuesta.text, "html.parser")

tablas = soup.find_all("table", class_="wikitable")

print("Tablas encontradas:", len(tablas))

df_web = pd.read_html(StringIO(str(tablas[0])))[0]

display(df_web.head())
```


##  7. Unir los DataFrames obtenidos  
Combina los distintos DataFrames en uno solo (`df_combinado`) usando `pd.concat()` o `pd.merge()`.


```python

df1 = df_vehicles.copy()
df1["fuente"] = "Vuelos"

df2 = df_bicicletas.copy()
df2["fuente"] = "Bicicletas"

df3 = df_auto.copy()
df3["fuente"] = "Automoviles"

df4 = df_web.copy()
df4["fuente"] = "Produccion de vehiculos"

df_combinado = pd.concat([df1, df2, df3, df4], ignore_index=True)

display(df_combinado.head())
```


##  8. Exploración básica del DataFrame combinado  
Usa funciones como `.head()`, `.info()`, `.describe()`, `.max()`, `.min()` para explorar los datos.


```python
display(df_combinado.head())

df_combinado.info()

display(df_combinado.describe())

display(df_combinado.max(numeric_only=True))

display(df_combinado.min(numeric_only=True))
```
